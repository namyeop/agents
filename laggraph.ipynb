```python
from langgraph.graph import StateGraph, START, END
from typing_extensions import TypedDict
```

```python
class State(TypedDict):
    count: int
```

```python
state_graph = StateGraph(State)

def node_start(state: State) -> list[State]:
  return

def node_end(state: State) -> list[State]:
  return

state_graph.add_node("start", node_start)
state_graph.add_node("end", node_end)

state_graph.add_edge(START, "start")
state_graph.add_edge("start", "end")
state_graph.add_edge("end", END)
```

```python
graph = state_graph.compile()
```

```python
graph.invoke({
    "count": {
      "$lt": 5
    }
})
```
